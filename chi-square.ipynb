```python
from pandas import read_csv, crosstab
from scipy.stats import chi2_contingency
```

```python
df = read_csv("airline_passenger_satisfaction.csv")
male_df = df[df["Gender"] == "Male"] # filter by gender
```

```python
# analyze on board service and class (Independence test-male passengers)
category1 = "Class"
category2 = "On-board Service"

# create contingency table 
crosstab = crosstab(male_df[category1], male_df[category2])
```

```python
crosstab
```

```python
# perform chi-square test
chi2, pval, degrees_of_freedom, expected_counts = chi2_contingency(crosstab)
```

```python
# Print the results
print(f"\nCategory1: {category1}, Category2: {category2} (Male Passengers)")
print(crosstab)
print(f"Chi-aquare: {chi2:.2f}, p-value: {pval:.4f}, DEGREES OF fREEDOM: {degrees_of_freedom}")
```

```python
# 2. Prime number
n = int(input("Enter number: "))

if n <= 1:
    print("Not Prime")  # 1 and numbers <= 0 are not prime
else:
    for i in range(2, int(n*0.5) + 1):  # Check divisors up to √n
        if n % i == 0:
            print("Not Prime")
            break
    else:
        print("Prime")
```

```python
# 3. Armstrong
n = input("Enter a number: ")
arm = 0
for i in n:
  arm+= int(i)**3

if arm == int(n):
    print("Armstrong")
else:
    print("Not Armstrong")
```

```python
# 5. fibonacci using generators
# allows you to produce values one at a time in a memory-efficient way
def fibonacci_generator(n):
    """Generate the first n numbers in the Fibonacci sequence."""
    a, b = 0, 1  # Starting values for Fibonacci
    for _ in range(n):
        yield a  # Yield the current value of 'a'
        a, b = b, a + b  # Update to the next Fibonacci numbers
n = int(input("Enter the number of Fibonacci terms to generate: "))

# Create the generator
fib_gen = fibonacci_generator(n)

# Print Fibonacci numbers
print("Fibonacci sequence:")
for num in fib_gen:
    print(num, end=" ")
```

```python
# 6. Fibonacci

def fib(n):
    l = [0, 1]
    if n <= 0:
        return "Invalid input"  # Fibonacci is undefined for negative numbers
    for i in range(2, n):
      l.append(l[i-1] + l[i-2])
    return l

# Test the function
print(fib(10))
```

```python
# 8. pallindrome
num = 1221
orgnNum = num
reverse = 0
while num > 0:
  reverse = (reverse * 10) + (num % 10)
  num //= 10

print(reverse==orgnNum)
```

```python
# add digits of a number using recursion
def add(num):
  if num==0:
    return 0
  return num%10 + add(num//10)

add(12345)
```

```python
# add without '+'
def add_without_plus(a, b):
    while b != 0:
        # XOR gives the sum without carry
        temp_sum = a ^ b
        # AND gives the carry, shift it left by 1
        carry = (a & b) << 1
        # Update a and b for the next iteration
        a = temp_sum
        b = carry
    return a

# Test the function
print(add_without_plus(5, 7))  # Output: 12
```

```python
# Subtract withoout '-'
def subtract_without_minus(a, b):
    # The idea is to add a and the two's complement of b
    return a + (~b + 1)

# Test the function
print(subtract_without_minus(10, 3))
```
